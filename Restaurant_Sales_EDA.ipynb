```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x='total_bill', y='tip')

plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
sns.barplot(data=df, x='day', y='tip')

plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

```python
sns.barplot(data=df, x='time', y='total_bill')

plt.title('Average Bill by Time')
plt.xlabel('Time')
plt.ylabel('Average Bill')
plt.show()
```

```python
sns.barplot(data=df, x='size', y='total_bill')

plt.title('Group Size vs Total Bill')
plt.xlabel('Group Size')
plt.ylabel('Average Total Bill')
plt.show()
```

```python
sns.barplot(data=df, x='smoker', y='tip')

plt.title('Smoker vs Non-Smoker Tips')
plt.xlabel('Smoker')
plt.ylabel('Average Tip')
plt.show()
```

```python
sns.barplot(data=df, x='sex', y='tip')

plt.title('Average Tip by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Tip')
plt.show()
```

```python
df.groupby('day')['total_bill'].sum().sort_values(ascending=False)
```

```python
df.groupby('day')['total_bill'].mean().sort_values(ascending=False)
```

```python
df.groupby('size')['tip'].mean().sort_values(ascending=False)
```

```python
df.groupby('time')['total_bill'].sum().sort_values(ascending=False)
```

```python
df.groupby('day').agg({
    'total_bill': ['sum', 'mean', 'count'],
    'tip': 'mean'
})
```

```python
df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100

df[['total_bill', 'tip', 'tip_percentage']].head()
```

```python
df['tip_percentage'].mean()
```

```python
print("Total Sales:", df['total_bill'].sum())

print("Average Bill:", df['total_bill'].mean())

print("Highest Bill:", df['total_bill'].max())

print("Average Tip:", df['tip'].mean())

print("Average Tip Percentage:", df['tip_percentage'].mean())
```

```python
sns.barplot(data=df, x='day', y='total_bill', estimator=sum)

plt.title('Total Sales by Day')
plt.xlabel('Day')
plt.ylabel('Total Sales')
plt.show()
```

# Restaurant Data Analysis (EDA)

## Project Overview

This project analyzes restaurant sales data
to understand customer spending behavior,
sales patterns, and tipping habits.

## Tools Used
- Python
- Pandas
- Seaborn
- Matplotlib
- Google Colab




## Key Findings

1. Saturday recorded the highest total sales.
2. Friday recorded the lowest total sales.
3. Sunday had the highest average bill.
4. Dinner generated higher total sales than Lunch.
5. The average tip percentage was 16.08%.




## Business Recommendations

1. Increase marketing campaigns on Fridays
to attract more customers.

2. Focus on maintaining high sales on Saturdays.

3. Create special lunch offers to increase
customer spending.

4. Maintain good service quality during dinner
because customers spend more on average.

5. Encourage customer loyalty through discounts
and special offers.




## Conclusion

The analysis helped us understand restaurant
sales and customer behavior.

Saturday had the highest total sales,
while Sunday had the highest average bill.

Dinner generated higher average bills
than Lunch.

The results can help the restaurant
improve its sales strategy and customer experience.
